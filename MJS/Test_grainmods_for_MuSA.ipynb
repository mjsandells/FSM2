## Checks FSM2 grain size modifications for MuSA have been applied correctly 
i.e. no difference from original FSM2.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
new_states = pd.read_csv('Alptal_new_stat.txt', sep=r"\s+", header=None)
```

```python
orig_states = pd.read_csv('Alptal_origstat.txt', sep=r"\s+", header=None)
```

```python
# Plot timeseries of last three columns
fig, (ax1, ax2) = plt.subplots(1, 2)
fig.suptitle('Layer grain sizes')
# Open site?
ax1.plot(1e3*new_states.iloc[:,-6:-3], linewidth=3,label=['top','middle','bottom'])
ax1.plot(1e3*orig_states.iloc[:,-6:-3], linewidth=1,label=['top','middle','bottom'])
# Veg site?
ax2.plot(1e3*new_states.iloc[:,-3:], linewidth=3,label=['new_top','new_middle','new_bottom'])
ax2.plot(1e3*orig_states.iloc[:,-6:-3], linewidth=1,label=['top','middle','bottom'])
#ax2.plot(1e3*orig_states.iloc[:,-3:], linewidth=1,label=['orig_top','orig_middle','orig_bottom'])
# Labels etc.
ax1.set_title('Open site')
ax2.set_title('Forest site')
ax1.set_xlabel('Time step')
ax2.set_xlabel('Time step')
ax1.set_ylabel('Grain radius in mm')
ax2.legend()
```

```python
# Check sequential nature of outputs: snow depth at two sites
plt.plot(new_states.iloc[:,4], linewidth=4, label='new, open')
plt.plot(new_states.iloc[:,5], linewidth=4, label='new, veg')
plt.plot(orig_states.iloc[:,4], linewidth=1, label='orig, open')
plt.plot(orig_states.iloc[:,5], linewidth=1, label='orig, veg')
plt.legend()
plt.xlabel('timestep')
plt.ylabel('snow depth (m)')
```
